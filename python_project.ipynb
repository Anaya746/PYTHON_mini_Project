```python

# ==========================================================
# PROJECT: E-COMMERCE PRODUCT REVIEW AND RATING CLEANER
# STUDENT: ANAYA MAAN
# COURSE: FIRST YEAR B.TECH
# TECHNOLOGIES: PYTHON, PANDAS, NUMPY
# DATASET: 100 VALID SAMPLE REVIEWS
# ==========================================================

# STEP 1: IMPORT LIBRARIES

import pandas as pd
import numpy as np

print("E-COMMERCE PRODUCT REVIEW AND RATING CLEANER")
print("Student Name: ANAYA MAAN")


# STEP 2: CREATE SAMPLE PRODUCT DATA

np.random.seed(42)

products = [
    "Laptop", "Mobile", "Headphones", "Keyboard",
    "Mouse", "Smart Watch", "Tablet", "Camera",
    "Speaker", "Monitor"
]

reviews = [
    "Very good product",
    "Excellent quality",
    "Good value for money",
    "Average product",
    "Amazing experience",
    "Not satisfied",
    "Nice product",
    "Worth buying",
    "Good performance",
    "Highly recommended"
]

ratings = [1, 2, 3, 4, 5]

data = []

for i in range(100):
    data.append({
        "Product_ID": 1001 + i,
        "Product_Name": products[i % 10],
        "Customer_Review": np.random.choice(reviews),
        "Rating": np.random.choice(ratings)
    })

df = pd.DataFrame(data)


# STEP 3: ADD SOME DATA ERRORS FOR CLEANING PRACTICE

# Add unnecessary spaces in some reviews

df.loc[5, "Customer_Review"] = "  " + df.loc[5, "Customer_Review"] + "  "
df.loc[15, "Product_Name"] = "  " + df.loc[15, "Product_Name"] + "  "

# Create two duplicate records

duplicates = df.iloc[[2, 15]].copy()

# Create one missing review record

missing_review = df.iloc[[20]].copy()
missing_review["Product_ID"] = 2001
missing_review["Customer_Review"] = None

# Create one missing rating record

missing_rating = df.iloc[[40]].copy()
missing_rating["Product_ID"] = 2002
missing_rating["Rating"] = np.nan

# Create two invalid rating records

invalid_ratings = df.iloc[[60, 80]].copy()
invalid_ratings["Product_ID"] = [2003, 2004]
invalid_ratings["Rating"] = [0, 6]

# Combine all records

df = pd.concat(
    [df, duplicates, missing_review,
     missing_rating, invalid_ratings],
    ignore_index=True
)


# STEP 4: DISPLAY ORIGINAL DATASET

print("\nORIGINAL DATASET")
print(df.head(15))

print("\nORIGINAL DATASET INFORMATION")
print(df.info())

print("\nMISSING VALUES")
print(df.isnull().sum())


# STEP 5: REMOVE DUPLICATE RECORDS

df = df.drop_duplicates()

print("\nAFTER REMOVING DUPLICATES")
print("Records remaining:", len(df))


# STEP 6: CLEAN TEXT DATA

df["Product_Name"] = df["Product_Name"].str.strip()

df["Customer_Review"] = df["Customer_Review"].str.strip()

print("\nTEXT DATA CLEANED SUCCESSFULLY")


# STEP 7: REMOVE MISSING VALUES

df = df.dropna(
    subset=["Customer_Review", "Rating"]
)

print("\nAFTER REMOVING MISSING VALUES")
print("Records remaining:", len(df))


# STEP 8: REMOVE INVALID RATINGS

df = df[
    (df["Rating"] >= 1) &
    (df["Rating"] <= 5)
]

df["Rating"] = df["Rating"].astype(int)

print("\nAFTER REMOVING INVALID RATINGS")
print("Records remaining:", len(df))


# STEP 9: DISPLAY FINAL CLEANED DATASET

print("\nFINAL CLEANED DATASET")
print(df.head(20))

print("\nFINAL DATASET INFORMATION")
print(df.info())


# STEP 10: CALCULATE AVERAGE RATING USING NUMPY

average_rating = np.mean(df["Rating"])

print("\nOVERALL AVERAGE RATING:",
      round(average_rating, 2))


# STEP 11: PRODUCT-WISE RATING ANALYSIS

product_summary = df.groupby("Product_Name").agg(
    Average_Rating=("Rating", "mean"),
    Total_Reviews=("Rating", "count")
).reset_index()

product_summary["Average_Rating"] = (
    product_summary["Average_Rating"].round(2)
)

print("\nPRODUCT-WISE RATING SUMMARY")
print(product_summary)


# STEP 12: FIND HIGHEST AND LOWEST RATED PRODUCTS

highest_product = product_summary.loc[
    product_summary["Average_Rating"].idxmax()
]

lowest_product = product_summary.loc[
    product_summary["Average_Rating"].idxmin()
]

print("\nHIGHEST RATED PRODUCT")
print(highest_product)

print("\nLOWEST RATED PRODUCT")
print(lowest_product)


# STEP 13: NUMPY STATISTICS

rating_array = df["Rating"].to_numpy()

print("\nRATING STATISTICS")
print("Maximum Rating:", np.max(rating_array))
print("Minimum Rating:", np.min(rating_array))
print("Average Rating:", np.mean(rating_array))
print("Median Rating:", np.median(rating_array))


# STEP 14: DATA CLEANING REPORT

original_records = 106
cleaned_records = len(df)

print("\nDATA CLEANING REPORT")
print("Original Records:", original_records)
print("Cleaned Records:", cleaned_records)
print("Records Removed:",
      original_records - cleaned_records)


# STEP 15: SAVE CLEANED DATASET

df.to_csv(
    "cleaned_product_reviews.csv",
    index=False
)

print("\nCleaned dataset saved successfully!")


# STEP 16: SAVE PRODUCT SUMMARY

product_summary.to_csv(
    "product_rating_summary.csv",
    index=False
)

print("Product summary saved successfully!")


# STEP 17: FINAL MESSAGE

print("\nPROJECT COMPLETED SUCCESSFULLY!")
print("Student Name: ANAYA MAAN")
print("Total Valid Reviews:", len(df))
```
