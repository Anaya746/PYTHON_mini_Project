```python

# ==========================================================
# PROJECT: E-COMMERCE PRODUCT REVIEW AND RATING CLEANER
# STUDENT NAME: ANAYA MAAN
# LEVEL: 1st YEAR B.TECH
# TECHNOLOGIES: PYTHON, PANDAS, NUMPY
# ==========================================================

# STEP 1: Import required libraries

import pandas as pd
import numpy as np

print("E-Commerce Product Review and Rating Cleaner")
print("Student Name: ANAYA MAAN")


# STEP 2: Create a sample dataset

data = {
    "Product_ID": [101, 102, 103, 104, 105, 101,
                   106, 107, 108, 109, 110, 111],

    "Product_Name": [
        "Laptop", "Mobile", "Headphones", "Keyboard",
        "Mouse", "Laptop", "Smart Watch", "Tablet",
        "Camera", "Speaker", "Charger", "Monitor"
    ],

    "Customer_Review": [
        "Very good product",
        "Excellent quality",
        "Good sound",
        "  Nice keyboard  ",
        "Value for money",
        "Very good product",
        None,
        "Average product",
        "Excellent camera",
        "Bad quality",
        "Useful product",
        "Very clear display"
    ],

    "Rating": [5, 4, 4, 5, 3, 5, 6, 3, 5, 0, np.nan, 4]
}

df = pd.DataFrame(data)

print("\nORIGINAL DATASET")
print(df)


# STEP 3: Display dataset information

print("\nDATASET INFORMATION")
print(df.info())

print("\nMISSING VALUES")
print(df.isnull().sum())


# STEP 4: Remove duplicate records

df = df.drop_duplicates()

print("\nDATA AFTER REMOVING DUPLICATES")
print(df)


# STEP 5: Clean text data

df["Product_Name"] = df["Product_Name"].str.strip()

df["Customer_Review"] = df["Customer_Review"].str.strip()

print("\nAFTER TEXT CLEANING")
print(df)


# STEP 6: Handle missing values

# Remove rows where review or rating is missing

df = df.dropna(subset=["Customer_Review", "Rating"])

print("\nAFTER REMOVING MISSING VALUES")
print(df)


# STEP 7: Remove invalid ratings

# Ratings must be between 1 and 5

df = df[(df["Rating"] >= 1) & (df["Rating"] <= 5)]

print("\nAFTER REMOVING INVALID RATINGS")
print(df)


# STEP 8: Convert ratings into integer values

df["Rating"] = df["Rating"].astype(int)

print("\nFINAL CLEANED DATASET")
print(df)


# STEP 9: Calculate average rating

average_rating = np.mean(df["Rating"])

print("\nOVERALL AVERAGE RATING:", round(average_rating, 2))


# STEP 10: Product-wise rating analysis

product_summary = df.groupby("Product_Name").agg(
    Average_Rating=("Rating", "mean"),
    Total_Reviews=("Rating", "count")
).reset_index()

print("\nPRODUCT-WISE RATING SUMMARY")
print(product_summary)


# STEP 11: Find highest rated product

highest_product = product_summary.loc[
    product_summary["Average_Rating"].idxmax()
]

print("\nHIGHEST RATED PRODUCT")
print(highest_product)


# STEP 12: Calculate rating statistics using NumPy

ratings = df["Rating"].to_numpy()

print("\nRATING STATISTICS")
print("Maximum Rating:", np.max(ratings))
print("Minimum Rating:", np.min(ratings))
print("Average Rating:", np.mean(ratings))
print("Median Rating:", np.median(ratings))


# STEP 13: Compare original and cleaned data

print("\nDATA CLEANING REPORT")
print("Original number of records:", len(data["Rating"]))
print("Cleaned number of records:", len(df))
print("Records removed:", len(data["Rating"]) - len(df))


# STEP 14: Save cleaned dataset

df.to_csv("cleaned_product_reviews.csv", index=False)

print("\nCleaned dataset saved successfully!")


# STEP 15: Save product summary

product_summary.to_csv("product_rating_summary.csv", index=False)

print("Product summary saved successfully!")

print("\nPROJECT COMPLETED SUCCESSFULLY!")
```
